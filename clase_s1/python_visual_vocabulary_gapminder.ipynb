## 1. Cargar librerias

### Que hace cada una

- `pandas`: tablas y transformaciones
- `matplotlib`: base del grafico
- `seaborn`: graficos mas faciles y mejor estilo
- `plotly.express`: interactividad


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 2. Cargar el dataset

Vamos a usar `gapminder`, que viene listo dentro de Plotly.

### Que significa cada columna importante

- `country`: pais
- `continent`: continente
- `year`: ano
- `lifeExp`: esperanza de vida
- `gdpPercap`: PIB per capita
- `pop`: poblacion


```python
df = px.data.gapminder().copy()
df.head()
```

```python
df.shape, df.columns.tolist()
```

## 3. Caso de clase

> Como ha cambiado el desarrollo humano entre 1952 y 2007, que regiones avanzan mas lento y donde deberiamos mirar primero?

Para varias preguntas usaremos una fotografia del ano 2007.

- `df` es la tabla completa
- `df_2007` sera una version filtrada


```python
year_selected = 2007
df_2007 = df[df['year'] == year_selected].copy()
df_2007.head()
```

## 4. Correlation

### Pregunta visual

> Hay relacion entre PIB per capita y esperanza de vida?

La familia visual mas natural aqui es `correlation`.

Los graficos sugeridos son:
- `scatter plot`
- `bubble chart`


### Matplotlib · Paso 1

Arrancamos con lo minimo: puntos en X y Y.

- `x` = `gdpPercap`
- `y` = `lifeExp`


```python
fig, ax = plt.subplots()
ax.scatter(df_2007['gdpPercap'], df_2007['lifeExp'])
plt.show()
```

### Matplotlib · Paso 2

Ahora hacemos el grafico legible.

- `alpha` controla transparencia
- `color` cambia el color
- `set_xscale('log')` pone el eje X en escala logaritmica
- `set_title`, `set_xlabel`, `set_ylabel` agregan contexto


```python
fig, ax = plt.subplots()
ax.scatter(df_2007['gdpPercap'], df_2007['lifeExp'], alpha=0.7, color='steelblue')
ax.set_xscale('log')
ax.set_title('Correlation · Matplotlib')
ax.set_xlabel('PIB per capita (log)')
ax.set_ylabel('Esperanza de vida')
plt.show()
```

### Seaborn · Paso 3

Ahora hacemos explicitas las esteticas.

- `data=df_2007` dice de donde salen los datos
- `x` y `y` indican ejes
- `hue='continent'` colorea por continente
- `size='pop'` cambia tamano por poblacion


```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_2007,
    x='gdpPercap',
    y='lifeExp',
    hue='continent',
    size='pop',
    sizes=(20, 300),
    alpha=0.7
)
plt.xscale('log')
plt.title('Correlation · Seaborn')
plt.xlabel('PIB per capita (log)')
plt.ylabel('Esperanza de vida')
plt.show()
```

### Plotly · Paso 4

Misma pregunta, pero ahora con interactividad.

- `hover_name='country'` muestra el pais al pasar el mouse
- `color` y `size` mantienen la logica anterior
- `log_x=True` conserva la escala del eje X


```python
fig = px.scatter(
    df_2007,
    x='gdpPercap',
    y='lifeExp',
    color='continent',
    size='pop',
    hover_name='country',
    log_x=True,
    title='Correlation · Plotly'
)
fig.show()
```

## 5. Ranking

### Pregunta visual

> Que paises estan arriba y abajo en esperanza de vida?

Los graficos sugeridos son:
- `barra horizontal ordenada`
- `lollipop chart`

Antes de dibujar, hay que ordenar.


```python
ranking = df_2007[['country', 'lifeExp']].sort_values('lifeExp', ascending=False).head(15)
ranking_sorted = ranking.sort_values('lifeExp')
ranking_sorted.head()
```

### Matplotlib

- `barh` crea barras horizontales
- usamos `ranking_sorted` para que el orden se lea mejor


```python
fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(ranking_sorted['country'], ranking_sorted['lifeExp'], color='#2F8373')
ax.set_title('Ranking · Matplotlib')
ax.set_xlabel('Esperanza de vida')
ax.set_ylabel('Pais')
plt.show()
```

### Seaborn

- `data=ranking_sorted` define la tabla
- `x='lifeExp'` pone el valor en horizontal
- `y='country'` deja el pais en vertical


```python
plt.figure(figsize=(10, 7))
sns.barplot(data=ranking_sorted, x='lifeExp', y='country', hue='country', palette='viridis', legend=False)
plt.title('Ranking · Seaborn')
plt.xlabel('Esperanza de vida')
plt.ylabel('Pais')
plt.show()
```

### Plotly

La logica visual no cambia; cambia la experiencia del lector.


```python
fig = px.bar(
    ranking_sorted,
    x='lifeExp',
    y='country',
    orientation='h',
    title='Ranking · Plotly'
)
fig.show()
```

## 6. Distribution

### Pregunta visual

> Como se distribuye la esperanza de vida? Hay diferencias por continenten

Los graficos sugeridos son:
- `histograma`
- `boxplot`


### Matplotlib

- `hist` crea el histograma
- `bins` controla el numero de grupos


```python
fig, ax = plt.subplots()
ax.hist(df_2007['lifeExp'], bins=20, color='tomato', edgecolor='white')
ax.set_title('Distribution · Matplotlib')
ax.set_xlabel('Esperanza de vida')
ax.set_ylabel('Frecuencia')
plt.show()
```

### Seaborn

Aqui usamos `boxplot` para comparar grupos.


```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_2007, x='continent', y='lifeExp')
plt.title('Distribution · Seaborn')
plt.xlabel('Continente')
plt.ylabel('Esperanza de vida')
plt.show()
```

### Plotly

La version interactiva deja explorar por continente.


```python
fig = px.histogram(
    df_2007,
    x='lifeExp',
    color='continent',
    nbins=20,
    title='Distribution · Plotly'
)
fig.show()
```

## 7. Change over Time

### Pregunta visual

> Como ha cambiado la esperanza de vida promedio por continente entre 1952 y 2007?

Los graficos sugeridos son:
- `line chart`
- `area chart`

Antes de graficar, primero resumimos.


```python
serie_tiempo = (
    df.groupby(['year', 'continent'], as_index=False)['lifeExp']
      .mean()
)
serie_tiempo.head()
```

### Matplotlib

Una linea por continente.


```python
fig, ax = plt.subplots()
for continente, grupo in serie_tiempo.groupby('continent'):
    ax.plot(grupo['year'], grupo['lifeExp'], marker='o', label=continente)
ax.set_title('Change over Time · Matplotlib')
ax.set_xlabel('Ano')
ax.set_ylabel('Esperanza de vida promedio')
ax.legend()
plt.show()
```

### Seaborn

Seaborn reduce bastante el codigo de la serie temporal.


```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=serie_tiempo, x='year', y='lifeExp', hue='continent', marker='o')
plt.title('Change over Time · Seaborn')
plt.xlabel('Ano')
plt.ylabel('Esperanza de vida promedio')
plt.show()
```

### Plotly

Mismo argumento visual, pero interactivo.


```python
fig = px.line(
    serie_tiempo,
    x='year',
    y='lifeExp',
    color='continent',
    markers=True,
    title='Change over Time · Plotly'
)
fig.show()
```

## 8. Nota sobre Flow

Gapminder no trae columnas listas como `source`, `target` y `value`.

Pero eso no significa que `flow` no se pueda trabajar.

### Como si puede construirse

Con este dataset puedes derivar transiciones, por ejemplo:
- cambio de ranking entre 1952 y 2007
- paso de una categoria de esperanza de vida a otra
- paso de una categoria de PIB per capita a otra

Es decir:

`Flow no viene nativo, pero si puede construirse si defines estados y transiciones.`
